# 02. Distance matrix

## 2-1. From scratch, build a distance function

```python
v = [3, 4]

dist = 0

for i in v:
    dist = dist + i**2

dist = dist**0.5
print(dist)
```

```python
p1 = [2, 5]
p2 = [6, 8]

dist = 0

for i in range(len(p1)):
    dist = dist + (p1[i] - p2[i])**2

dist = dist**0.5
print(dist)
```

```python
def distance(p1, p2):
    dist = 0
    for i in range(len(p1)):
        dist = dist + (p1[i] - p2[i])**2
    dist = dist**0.5
    return dist
```

```python
distance(p1, p2)
```

## 2-2. np.linalg.norm

```python
v = np.array([3, 4])
length = np.linalg.norm(v)   # √(3² + 4²) = 5
print(length)                # 5.0
```

```python
p1 = np.array([2, 5])
p2 = np.array([6, 8])
distance = np.linalg.norm(p1 - p2)  # √((6-2)² + (8-5)²)
print(distance)                      # 5.0
```

```python
points = np.array([[0,0], [1,1], [2,2]])
centroids = np.array([[0,1], [2,0]])

# distances[i, j] = distance from points[i] to centroids[j]
# numpy broadcast
distances = np.linalg.norm(points[:, None] - centroids[None, :], axis=2)
print(distances)
```

# 03. Lables (np.argmin)

`axis=1` → column index of min in each row.

`axis=0` → row index of min in each column.

```python
arr = np.array([8, 3, 5, 1])
idx = np.argmin(arr)   # position of the smallest value

print(idx)             # 3
print(arr[idx])        # 1
```

```python
matrix = np.array([[10, 2, 8],
                   [ 5, 7, 1]])

# column index of the minimum in each row
col_idx = np.argmin(matrix, axis=1)
print(col_idx)         # [1 2]
```
