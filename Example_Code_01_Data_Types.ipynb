## 1-5. Visualize the top 5 directors with the highest average movie rating.

```python
director_rating = imdb_df.groupby("Director")["Rating"].mean()
top5_directors = director_rating.sort_values(ascending=False).head(5)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(top5_directors.index, top5_directors.values)

plt.title("Top 5 Directors by Average Movie Rating")
plt.xlabel("Director")
plt.ylabel("Average Rating")
plt.ylim(8, 9)
plt.xticks(rotation=45)

plt.show()
```

## 2-8. Do the three Iris species have significantly different petal lengths?

```python
from scipy.stats import f_oneway
```

```python
setosa = iris_df[iris_df["species"] == "setosa"]["petal_length"]
versicolor = iris_df[iris_df["species"] == "versicolor"]["petal_length"]
virginica = iris_df[iris_df["species"] == "virginica"]["petal_length"]

f_stat, p_value = f_oneway(setosa, versicolor, virginica)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

## 2-12. Visualize Petal Length and Width

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 8]

plt.scatter(x, y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Plot")

plt.show()
```

```python
setosa = iris_df[iris_df["species"] == "setosa"]
versicolor = iris_df[iris_df["species"] == "versicolor"]
virginica = iris_df[iris_df["species"] == "virginica"]

plt.scatter(setosa["petal_length"],setosa["petal_width"],label="Setosa")
plt.scatter(versicolor["petal_length"],versicolor["petal_width"],label="Versicolor")
plt.scatter(virginica["petal_length"],virginica["petal_width"],label="Virginica")

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Petal Length vs. Petal Width")
plt.legend()

plt.show()
```

```python
species_list = iris_df["species"].unique()

for species in species_list:
    data = iris_df[iris_df["species"] == species]

    plt.scatter(
        data["petal_length"],
        data["petal_width"],
        label=species
    )

plt.title("Petal Length vs. Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend()

plt.show()
```
